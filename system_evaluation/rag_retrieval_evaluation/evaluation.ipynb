```python
from deepeval import evaluate
from deepeval.test_case import LLMTestCase

from deepeval.metrics import (
    ContextualRelevancyMetric,
    ContextualRecallMetric,
    ContextualPrecisionMetric,
)

from rag_system.rag_functions import retrieve_semantic_chunks
from system_evaluation.rag_retrieval_evaluation.eval_data import eval_data

from dotenv import load_dotenv
from deepeval.models import OllamaModel , GeminiModel

from deepeval.evaluate import AsyncConfig, DisplayConfig

import os

load_dotenv()


ollama  = OllamaModel(
    model='llama3.2:latest',
    base_url="localhost:11434"
)

os.environ["DEEPEVAL_DISABLE_TIMEOUTS"] = "1" 

user_id = "c135f300-3fbb-4eec-8f73-061bcf63cb40"
```

```python
test_cases = []
print(f"Total evaluation cases: {len(eval_data)}")

for index, item in enumerate(eval_data, start=1):

    query = item["query"]

    print("\n")
    print("-" * 80)
    print(f"TEST CASE {index}/{len(eval_data)}")

    docs, _ = retrieve_semantic_chunks(
            pillar=query,
            user_id=user_id
        )

    retrieval_context = [
            doc.page_content
            for doc in docs
        ]


    print(f"Retrieved chunks: {len(retrieval_context)}")



    expected_contexts = item["expected_contexts"]


    expected_output = "\n\n".join(
            expected_contexts
        )

    actual_output = "\n\n".join(
            retrieval_context
        )

    test_case = LLMTestCase(

            # The pillar supplied to the retriever
            input=query,

            # What the retriever actually returned
            actual_output=actual_output,

            # What the ideal retrieval should contain
            expected_output=expected_output,

            # Actual retrieved chunks
            retrieval_context=retrieval_context,

        )


    test_cases.append(test_case)
```

```python
test_cases
```

```python
precision = ContextualPrecisionMetric(
    model=ollama,
    threshold=0.7
)

relavance = ContextualRelevancyMetric(
    model=ollama,
    threshold=0.7
)
```

```python
result = evaluate(
        test_cases=test_cases,
        metrics=[
            relavance,
        ],
        async_config=AsyncConfig(run_async=False),
)
```

```python
result = evaluate(
        test_cases=test_cases,
        metrics=[
            precision,
        ],
        async_config=AsyncConfig(run_async=False),
)
```
